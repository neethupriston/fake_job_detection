```python
from gensim.models import KeyedVectors
from gensim.scripts.glove2word2vec import glove2word2vec

raw_glove_file = 'glove.6B.100d.txt'
word2vec_file = 'glove.6B.100d.word2vec.txt'
```

```python

# Check if the converted Word2Vec file already exists
if not os.path.exists(word2vec_file):
    print(f"'{word2vec_file}' not found. Checking for the raw '{raw_glove_file}'...")

    # If the raw file exists, convert it
    if os.path.exists(raw_glove_file):
        print("Raw file found! Converting to Word2Vec format (this may take a minute)...")
        glove2word2vec(raw_glove_file, word2vec_file)
        print("Conversion complete!")
    else:
        # If neither file exists, stop the code and warn the user
        raise FileNotFoundError(f"Missing File! Please upload '{raw_glove_file}' to your current folder.")

# Load the model into memory

glove_model = KeyedVectors.load_word2vec_format(word2vec_file)


# ==========================================
# STEP 3: CATEGORIZE & ENCODE
# ==========================================
# Extract unique categories (ignoring missing/NaN values)
unique_categories_train = X_train_fake_job['industry'].unique()
print(f"Found {len(unique_categories_train)} unique categories to encode.")

def get_mean_glove_vector(text, model):
    """Splits a category name into words and averages their GloVe vectors."""
    words = str(text).lower().split()
    vectors = [model[w] for w in words if w in model]
    #print(model.vector_size)
    if not vectors:
        return np.zeros(model.vector_size) # Return array of zeros if no words match

    return np.mean(vectors, axis=0)


# Create a dictionary mapping: { 'Industry Name' : [Vector Array] }
category_to_vector = {cat: get_mean_glove_vector(cat, glove_model) for cat in unique_categories_train}


# ==========================================
# STEP 4: MAP BACK TO MAIN DATAFRAME
# ==========================================
# Map the vectors back to the original dataset based on the industry name
X_train_fake_job['industry_map'] = X_train_fake_job['industry'].map(category_to_vector)

print("\n--- Final Dataset Preview ---")
print(type(X_train_fake_job['industry_map']))
#print(df[['industry', 'industry_embedded']].head())
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from transformers import RobertaModel
from torch.optim import AdamW
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from tqdm import tqdm # For progress bars
import numpy as np
print("Is CUDA available?:", torch.cuda.is_available())
```

```python

# ==========================================
# 1. DATASET CLASS
# ==========================================
class JobFraudDataset(Dataset):
    def __init__(self, roberta_ids, roberta_masks, glove_data,remaininput_data,numeric_data, labels):
        self.roberta_ids = roberta_ids
        self.roberta_masks = roberta_masks
        self.glove_data = glove_data
        self.numeric_data = numeric_data
        self.labels = labels
        self.remaininput_data = remaininput_data
    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return {
            'roberta_ids': self.roberta_ids[idx].clone().detach().long(),
            'roberta_mask': self.roberta_masks[idx].clone().detach().long(),
            # Convert the already embedded GloVe vectors directly to float32
            'glove_data': torch.tensor(self.glove_data[idx], dtype=torch.float32),
            'numeric_data': torch.tensor(self.numeric_data[idx], dtype=torch.float32),
             'remaininput_data': torch.tensor(self.remaininput_data[idx],dtype=torch.float32),
            'label': torch.tensor(self.labels[idx], dtype=torch.long)

            }

# ==========================================
# 2. THE HYBRID MODEL
# ==========================================
class UltimateJobFraudClassifier(nn.Module):
    def __init__(self, glove_dim, lstm_hidden, num_numeric_features, num_classes):
        super(UltimateJobFraudClassifier, self).__init__()

        self.roberta = RobertaModel.from_pretrained("roberta-base")

        # NO EMBEDDING LAYER: glove_data is already embedded (float vectors)
        # We pass it directly to the LSTM
        self.bi_lstm = nn.LSTM(input_size=glove_dim, hidden_size=lstm_hidden, batch_first=True, bidirectional=True)

        self.numeric_branch = nn.Sequential(
            nn.Linear(num_numeric_features, 16),
            nn.ReLU()
        )

        combined_dim = 768 + (lstm_hidden * 2) + 16
        self.classifier = nn.Sequential(
            nn.Linear(combined_dim, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256,64),
            nn.ReLU(),
            nn.Linear(64, num_classes)
        )

    def forward(self, roberta_ids, roberta_mask, glove_data, numeric_data,remaininput_data):
        roberta_features = self.roberta(input_ids=roberta_ids, attention_mask=roberta_mask).pooler_output

        # Pass glove_data directly to LSTM
        lstm_out, (hn, _) = self.bi_lstm(glove_data)
        glove_features = torch.cat((hn[-2, :, :], hn[-1, :, :]), dim=1)

        combined_numeric = torch.cat((numeric_data, remaininput_data), dim=1)
        numeric_features = self.numeric_branch(combined_numeric)

        final_combined = torch.cat((roberta_features, glove_features, numeric_features), dim=1)
        return self.classifier(final_combined)

# ==========================================
# 3. DATA PREPARATION (Combining Tensors)
# ==========================================
print("Combining Data...")

print("Properly Stacking GloVe Vectors...")

glove_cols = [
    'industry_map',
    'employment_type_encoded',
    'required_experience_encoded',
    'required_education_encoded',
    'function_encoded'
]

# 1. Function to ensure every item is exactly 100 numbers long
GLOVE_DIM = 100 # Based on your error, your GloVe vectors are 100D

def fix_glove_vector(vec):
    # If it is a valid vector of 100 numbers, keep it
    if isinstance(vec, (list, np.ndarray)) and len(vec) == GLOVE_DIM:
        return np.array(vec, dtype=np.float32)
    # If it is a single number, missing, or broken, return 100 zeros
    else:
        return np.zeros(GLOVE_DIM, dtype=np.float32)

# 2. Apply the fix to all your GloVe columns in Train and Test data
for col in glove_cols:
    X_train_fake_job[col] = X_train_fake_job[col].apply(fix_glove_vector)
    X_test_fake_job[col] = X_test_fake_job[col].apply(fix_glove_vector)

# 3. Now stack them safely! Every vector is guaranteed to be size 100.
X_train_glove = np.array([np.vstack(row) for row in X_train_fake_job[glove_cols].values], dtype=np.float32)
X_test_glove = np.array([np.vstack(row) for row in X_test_fake_job[glove_cols].values], dtype=np.float32)

print(f"Train GloVe Shape: {X_train_glove.shape}") # Should successfully print (Number_of_Jobs, 5, 100)


# Combine Train RoBERTa Data (Max 512 tokens)
X_train_roberta_ids = torch.cat((
    title_encoded_batch['input_ids'], company_profile_encoded_batch['input_ids'],
    encoded_batch_description['input_ids'], requirements_encoded_batch['input_ids'],
    benefits_encoded_batch['input_ids'], department_encoded_batch['input_ids'],
    location_encoded_batch_train['input_ids']
), dim=1)[:, :512]

X_train_roberta_masks = torch.cat((
    title_encoded_batch['attention_mask'], company_profile_encoded_batch['attention_mask'],
    encoded_batch_description['attention_mask'], requirements_encoded_batch['attention_mask'],
    benefits_encoded_batch['attention_mask'], department_encoded_batch['attention_mask'],
    location_encoded_batch_train['attention_mask']
), dim=1)[:, :512]

# Combine Test RoBERTa Data (Max 512 tokens) - Fixed the attention_mask typo here!
X_test_roberta_ids = torch.cat((
    title_encoded_batch_test['input_ids'], company_profile_encoded_batch_test['input_ids'],
    encoded_batch_description_test['input_ids'], requirements_encoded_batch_test['input_ids'],
    benefits_encoded_batch_test['input_ids'], department_encoded_batch_test['input_ids'],
    location_encoded_batch_test['input_ids']
), dim=1)[:, :512]

X_test_roberta_masks = torch.cat((
    title_encoded_batch_test['attention_mask'], company_profile_encoded_batch_test['attention_mask'],
    encoded_batch_description_test['attention_mask'], requirements_encoded_batch_test['attention_mask'],
    benefits_encoded_batch_test['attention_mask'], department_encoded_batch_test['attention_mask'],
    location_encoded_batch_test['attention_mask']
), dim=1)[:, :512]
X_train_scaled = X_train_fake_job[['min_salary_scaled', 'max_salary_scaled']].values
X_test_scaled = X_test_fake_job[['min_salary_scaled', 'max_salary_scaled']].values
X_train_datar = X_train_fake_job[['telecommuting', 'has_company_logo', 'has_questions']].values
X_test_datar = X_test_fake_job[['telecommuting', 'has_company_logo', 'has_questions']].values
X_train_scaled = np.nan_to_num(np.array(X_train_scaled, dtype=np.float32))
X_test_scaled = np.nan_to_num(np.array(X_test_scaled, dtype=np.float32))
X_train_datar = np.nan_to_num(np.array(X_train_datar, dtype=np.float32))
X_test_datar = np.nan_to_num(np.array(X_test_datar, dtype=np.float32))

# Create DataLoaders
train_dataset = JobFraudDataset(X_train_roberta_ids, X_train_roberta_masks, X_train_glove,X_train_datar, X_train_scaled, y_train_fake_job.values)
test_dataset = JobFraudDataset(X_test_roberta_ids, X_test_roberta_masks, X_test_glove,X_test_datar, X_test_scaled, y_test_fake_job.values)


train_dataloader = DataLoader(train_dataset, batch_size=64, shuffle=True,num_workers=4)
test_dataloader = DataLoader(test_dataset, batch_size=64, shuffle=False,num_workers=4)

# ==========================================
# 4. TRAINING SETUP
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

model = UltimateJobFraudClassifier(
    glove_dim=100, lstm_hidden=128, num_numeric_features=5, num_classes=2).to(device)

# Handling Imbalance (1:19 ratio for 95% Real, 5% Fake)
class_weights = torch.tensor([5.0,15.0]).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights)

# Optimizer (AdamW is best for RoBERTa)
optimizer = AdamW(model.parameters(), lr=2e-5) # Very small learning rate for Transformers

# ==========================================
# 5. THE TRAINING LOOP
# ==========================================
EPOCHS = 5


scaler = torch.cuda.amp.GradScaler()

print("\n--- Starting Training ---")
for epoch in range(EPOCHS):
    model.train() # Set model to training mode
    total_loss = 0

    # tqdm adds a progress bar
    progress_bar = tqdm(train_dataloader, desc=f"Epoch {epoch+1}/{EPOCHS}")

    for batch in progress_bar:
        # Move data to GPU/CPU
        b_roberta_ids = batch['roberta_ids'].to(device)
        b_roberta_mask = batch['roberta_mask'].to(device)
        b_glove = batch['glove_data'].to(device)
        b_numeric = batch['numeric_data'].to(device)
        b_remaininput = batch['remaininput_data'].to(device)
        b_labels = batch['label'].to(device)

        # 1. Clear previous gradients
        optimizer.zero_grad()


        with torch.cuda.amp.autocast():
            outputs = model(b_roberta_ids, b_roberta_mask, b_glove, b_numeric, b_remaininput)
            loss = criterion(outputs, b_labels)

        total_loss += loss.item()


        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        progress_bar.set_postfix({'Loss': f"{loss.item():.4f}"})

    avg_train_loss = total_loss / len(train_dataloader)
    print(f"\nAverage Training Loss for Epoch {epoch+1}: {avg_train_loss:.4f}\n")
# ==========================================
# 6. EVALUATION / PREDICTION LOOP
# ==========================================
print("\n--- Starting Evaluation on Test Data ---")
model.eval() # Set model to evaluation mode
all_preds, all_labels = [], []

with torch.no_grad():
    for batch in tqdm(test_dataloader, desc="Predicting"):
        b_roberta_ids = batch['roberta_ids'].to(device)
        b_roberta_mask = batch['roberta_mask'].to(device)
        b_glove = batch['glove_data'].to(device)
        b_numeric = batch['numeric_data'].to(device)
        b_remaininput = batch['remaininput_data'].to(device)
        b_labels = batch['label'].to(device)

        outputs = model(b_roberta_ids, b_roberta_mask, b_glove, b_numeric,b_remaininput)
        preds = torch.argmax(outputs, dim=1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(b_labels.cpu().numpy())

print("\n=== FINAL RESULTS ===")
print(f"Overall Accuracy: {accuracy_score(all_labels, all_preds) * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(all_labels, all_preds, target_names=['Real (0)', 'Fake (1)']))
print("Confusion Matrix:")
print(confusion_matrix(all_labels, all_preds))
```
